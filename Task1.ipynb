```python
import cv2
img1= cv2.imread("flowers.jpg",cv2.COLOR_BGR2RGB)
img2= cv2.imread("praying mantis.png",cv2.COLOR_BGR2RGB)
print(img1.shape)
print(img2.shape)
```

```python
import matplotlib.pyplot as plt
plt.imshow(img1, cmap='gray')
plt.axis('off')
plt.show()
```

```python

plt.imshow(img2, cmap='gray')
plt.axis('off')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Resize to 300x300 pixels (Note: cv2.resize takes (width, height))
img1_resized = cv2.resize(img1, (450,300))
img2_resized = cv2.resize(img2, (300,450))

# Display side by side
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(img1_resized, cmap='gray')
axes[0].set_title("Resized Flowers (450x300)")
axes[0].axis('off')

axes[1].imshow(img2_resized, cmap='gray')
axes[1].set_title("Resized Mantis (300x450)")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
